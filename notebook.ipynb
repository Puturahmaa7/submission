## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

#### Identifying Potential Outliers Using the IQR Method

Mengidentifikasi kandidat outlier pada variabel numerik menggunakan metode Interquartile Range (IQR). Metode ini menentukan batas bawah dan batas atas berdasarkan kuartil pertama (Q1), kuartil ketiga (Q3), serta rentang antarkuartil (IQR). Observasi yang berada di luar batas tersebut ditandai sebagai kandidat outlier untuk pemeriksaan lebih lanjut.

```python
NUMERIC_VARIABLES = [
    "PM2.5", "PM10", "SO2", "NO2", "CO", "O3",
    "TEMP", "PRES", "DEWP", "RAIN", "WSPM"
]

def iqr_calculate(data, variables, group_by=None):
    results = []

    groups = data.groupby(group_by) if group_by else [("Global", data)]

    for name, group in groups:
        for col in variables:
            Q1 = group[col].quantile(0.25)
            Q3 = group[col].quantile(0.75)
            IQR = Q3 - Q1

            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR

            values = group[col].dropna()
            outliers = values[(values < lower) | (values > upper)]

            results.append({
                "station": name if group_by else "Global",
                "variabel": col,
                "jumlah_data_valid": len(values),
                "kuartil_1": Q1,
                "kuartil_3": Q3,
                "rentang_iqr": IQR,
                "batas_bawah": lower,
                "batas_atas": upper,
                "jumlah_kandidat_outlier": len(outliers),
                "persentase_kandidat_outlier": len(outliers) / len(values) * 100
            })

    return pd.DataFrame(results)
```

```python
global_outlier = iqr_calculate(
    air_df,
    NUMERIC_VARIABLES
)

display(global_outlier.round(2))
```

Berdasarkan hasil identifikasi kandidat outlier menggunakan metode IQR secara global, ditemukan bahwa jumlah kandidat outlier berbeda pada setiap variabel.

- SO2 memiliki persentase kandidat outlier tertinggi, yaitu 8,64% (35.566 data), diikuti CO sebesar 7,01% (28.054 data).
- WSPM memiliki 23.079 kandidat outlier (5,49%), sedangkan PM2.5 memiliki 19.142 kandidat (4,65%).
- PM10 dan O3 masing-masing memiliki kandidat outlier sebesar 3,54% dan 4,07%.
- NO2 memiliki persentase kandidat outlier relatif rendah, yaitu 1,72%.
- TEMP, PRES, dan DEWP tidak menunjukkan kandidat outlier berdasarkan batas IQR.
- RAIN memiliki 16.520 kandidat outlier (3,93%). Hal ini terjadi karena Q1 dan Q3 bernilai 0 sehingga IQR = 0. Akibatnya, setiap nilai hujan di atas 0 teridentifikasi sebagai kandidat outlier.

Hasil metode IQR menunjukkan bahwa kandidat outlier paling banyak ditemukan pada variabel SO2, CO, dan WSPM. Namun, hasil ini belum menunjukkan bahwa seluruh observasi tersebut merupakan kesalahan pengukuran. Konsentrasi polutan dan kondisi meteorologis dapat memiliki distribusi yang menceng serta variasi ekstrem yang memang terjadi secara alami. Oleh karena itu, kandidat outlier perlu diperiksa lebih lanjut berdasarkan rentang fisik, konsistensi antarvariabel, pola temporal, dan karakteristik masing-masing stasiun sebelum menentukan perlakuan terhadap data.

Hasil identifikasi menunjukkan bahwa jumlah kandidat outlier bervariasi antarvariabel dan stasiun. Tabel ini digunakan untuk meninjau kandidat berdasarkan batas IQR pada masing-masing stasiun. #### Fixing Missing Values Using Time-Based Interpolation

Menangani nilai yang hilang (missing values) pada variabel numerik menggunakan metode time-based interpolation. Metode ini memperkirakan nilai kosong berdasarkan pola perubahan data terhadap waktu pada masing-masing stasiun pemantauan. Interpolasi hanya diterapkan pada celah data (missing gaps) dengan durasi maksimal 24 jam dan dibatasi pada celah yang memiliki nilai valid sebelum dan sesudahnya (inside interpolation).

```python
MAX_INTERPOLATION_GAP_HOURS = 24

def interpolate_gaps(data, variables, max_gap_hours=MAX_INTERPOLATION_GAP_HOURS):

    result = data[variables].copy()

    for _, group in data.groupby("station", observed=False):
        group = group.sort_values("datetime")
        original_index = group.index
        group = group.set_index("datetime")

        for col in variables:
            series = group[col]
            interpolated = series.interpolate(
                method="time",
                limit_area="inside"
            )

            gap = series.isna()
            gap_id = gap.ne(gap.shift()).cumsum()
            gap_length = gap.groupby(gap_id).transform("sum")

            eligible = gap & (gap_length <= max_gap_hours)

            result.loc[original_index[eligible.values], col] = (
                interpolated[eligible].values
            )

    return result
```

```python
TARGET_CLEANING = [
    "PM2.5", "PM10", "TEMP",
    "PRES", "DEWP", "WSPM"
]

cleaning_air_df = (
    cleaning_air_df
    .sort_values(["station", "datetime"])
    .reset_index(drop=True)
)

cleaning_air_df[TARGET_CLEANING] = interpolate_gaps(
    cleaning_air_df,
    TARGET_CLEANING
)
```

```python
missing_evaluation = pd.DataFrame({
    "jumlah_data_hilang_sebelum": air_df[TARGET_CLEANING].isna().sum(),
    "jumlah_data_hilang_sesudah": cleaning_air_df[TARGET_CLEANING].isna().sum()
})

missing_evaluation["jumlah_data_berhasil_diinterpolasi"] = (
    missing_evaluation["jumlah_data_hilang_sebelum"]
    - missing_evaluation["jumlah_data_hilang_sesudah"]
)

display(missing_evaluation)
```

Berdasarkan hasil interpolasi berbasis waktu, jumlah missing values pada PM2.5 berkurang dari 8.739 menjadi 2.435, sedangkan PM10 berkurang dari 6.449 menjadi 1.629. Pada variabel meteorologis, seluruh missing values pada TEMP, PRES, DEWP, dan WSPM berhasil diisi karena memenuhi kriteria interpolasi yang ditetapkan, yaitu memiliki celah data maksimal 24 jam dengan nilai valid sebelum dan sesudahnya. Secara keseluruhan, interpolasi berhasil mengisi 12.636 nilai yang hilang pada keenam variabel numerik tersebut. Sementara itu, nilai kosong yang tidak memenuhi kriteria interpolasi tetap dipertahankan untuk menghindari estimasi yang kurang dapat diandalkan.

#### Fixing Missing Values Using Forward Fill

Menangani nilai yang hilang (missing values) pada variabel RAIN menggunakan metode Forward Fill (ffill). Pengisian dilakukan berdasarkan nilai valid sebelumnya pada masing-masing stasiun pemantauan dengan batas maksimal 6 jam berturut-turut. Metode ini diterapkan dengan mempertimbangkan karakteristik data curah hujan yang didominasi oleh nilai nol, sehingga nilai sebelumnya digunakan sebagai estimasi untuk mengisi celah data pendek. Nilai kosong yang melebihi batas tersebut tetap dipertahankan.

```python
cleaning_air_df["RAIN"] = (
    cleaning_air_df
    .groupby("station", observed=False)["RAIN"]
    .transform(lambda group: group.ffill(limit=6))
)
```

Sebanyak 390 missing values pada variabel RAIN berhasil diisi menggunakan metode Forward Fill (ffill) dengan batas maksimal enam jam berturut-turut pada stasiun yang sama. Pengisian memanfaatkan nilai valid sebelumnya dan seluruh gap memenuhi kriteria pengisian yang ditetapkan.

#### Calculating Daily Average PM2.5 for Each Monitoring Station

Menghitung rata-rata konsentrasi PM2.5 harian berdasarkan kombinasi stasiun pemantauan (station) dan tanggal lengkap (tanggal). Kolom valid_hours menunjukkan jumlah pengamatan PM2.5 yang valid dalam satu hari pada masing-masing stasiun. Pengelompokan menggunakan tanggal lengkap, termasuk tahun, bulan, dan hari, sehingga pengamatan pada tanggal yang sama di bulan atau tahun berbeda tidak tercampur.

```python
daily_pm25 = (
    station.groupby(["station", "tanggal"], observed=True)["PM2.5"]
    .agg(
        daily_mean="mean",
        valid_hours="count"
    )
    .reset_index()
)

display(daily_pm25.head())
```

Hasil agregasi menunjukkan rata-rata PM2.5 harian (daily_mean) dan jumlah jam pengamatan valid (valid_hours) untuk setiap kombinasi stasiun dan tanggal. Lima baris pertama memperlihatkan data Stasiun Aotizhongxin pada 1–5 Maret 2013, dengan masing-masing 24 jam pengamatan valid. Agregasi ini memungkinkan perbandingan konsentrasi PM2.5 harian antarstasiun sekaligus mempertimbangkan kelengkapan pengamatan setiap hari.

#### Defining Valid-Day Criteria and the High-Concentration Threshold

Rata-rata PM2.5 harian hanya dihitung apabila memiliki minimal 18 jam pengamatan valid dari total 24 jam dalam sehari. Kriteria ini digunakan untuk memastikan bahwa rata-rata harian cukup merepresentasikan kondisi konsentrasi PM2.5 selama satu hari. Selanjutnya, ambang batas 75 µg/m³ ditetapkan sebagai kriteria operasional dalam proyek untuk mengidentifikasi hari dengan konsentrasi PM2.5 tinggi. Hari dengan rata-rata PM2.5 harian lebih besar dari 75 µg/m³ dikategorikan sebagai high-concentration days. Ambang tersebut merupakan nilai yang ditetapkan langsung dalam kode melalui PM25_THRESHOLD = 75, bukan nilai yang dihitung dari dataset pada tahap ini.

```python
MIN_VALID_HOURS = 18
PM25_THRESHOLD = 75

daily_valid = daily_pm25.loc[
    daily_pm25["valid_hours"] >= MIN_VALID_HOURS
].copy()

daily_valid["high_day"] = daily_valid["daily_mean"] > PM25_THRESHOLD

display(daily_valid.head())
```

Setelah menerapkan kriteria minimal 18 jam pengamatan valid per hari, data harian yang memenuhi syarat digunakan untuk mengidentifikasi hari dengan konsentrasi PM2.5 tinggi. Hari dikategorikan sebagai high_day apabila rata-rata PM2.5 harian melebihi 75 µg/m³. Sebagai contoh, Aotizhongxin pada 3 dan 5 Maret 2013 termasuk kategori hari dengan PM2.5 tinggi. Pendekatan ini membantu mengurangi bias akibat ketidaklengkapan pengamatan harian.

#### Calculating the Number and Percentage of High-PM2.5 Days per Station

Menghitung jumlah hari valid dan hari dengan rata-rata PM2.5 lebih besar dari 75 µg/m³ pada setiap stasiun. Persentase hari tinggi dihitung terhadap total hari valid di stasiun yang sama.

```python
high_day_summary = (
    daily_valid.groupby("station", observed=True)
    .agg(
        hari_valid=("tanggal", "count"),
        hari_tinggi=("high_day", "sum")
    )
)

high_day_summary["pct_hari_tinggi"] = (
    high_day_summary["hari_tinggi"]
    / high_day_summary["hari_valid"]
    * 100
)

display(high_day_summary.round(2))
```

Berdasarkan jumlah hari valid, persentase hari dengan rata-rata PM2.5 di atas 75 µg/m³ tertinggi tercatat di Dongsi sebesar 43,26%, diikuti Gucheng sebesar 42,93%, Nongzhanguan sebesar 42,87%, Wanshouxigong sebesar 42,86%, dan Wanliu sebesar 42,85%. Sementara itu, Dingling memiliki persentase terendah sebesar 31,91%, disusul Huairou sebesar 34,77% dan Changping sebesar 36,06%. Hasil ini menunjukkan adanya perbedaan frekuensi observasi harian dengan konsentrasi PM2.5 tinggi antarstasiun selama periode pengamatan.

#### Combining Station Statistics and Calculating Rankings

Menggabungkan jumlah dan persentase hari tinggi dengan statistik rata-rata PM2.5 setiap stasiun. Selanjutnya, menentukan peringkat stasiun berdasarkan rata-rata PM2.5 dan persentase hari tinggi, dari nilai tertinggi hingga terendah.

```python
station_stats = station_stats.join(high_day_summary, how="left")

station_stats["rank_mean"] = station_stats["mean"].rank(
    ascending=False,
    method="min"
).astype("Int64")

station_stats["rank_high_day"] = station_stats["pct_hari_tinggi"].rank(
    ascending=False,
    method="min"
).astype("Int64")
```

#### Calculating the Combined PM2.5 Index

Menghitung indeks gabungan berdasarkan dua indikator, yaitu rata-rata PM2.5 dan persentase hari tinggi. Masing-masing indikator dinormalisasi terhadap rata-rata seluruh stasiun, kemudian kedua indeks dirata-ratakan untuk menghasilkan indeks gabungan.

```python
average_station_high_day_pct = station_stats["pct_hari_tinggi"].mean()

station_stats["idx_mean"] = (
    station_stats["mean"] / overall_mean * 100
)

station_stats["idx_high_pct"] = (
    station_stats["pct_hari_tinggi"]
    / average_station_high_day_pct
    * 100
)

station_stats["indeks_gabungan"] = (
    station_stats["idx_mean"] + station_stats["idx_high_pct"]
) / 2

station_stats["rank_gabungan"] = (
    station_stats["indeks_gabungan"]
    .rank(ascending=False, method="min")
    .astype("Int64")
)
```

#### Identifying Stations with the Highest PM2.5 Levels

Mengidentifikasi stasiun dengan rata-rata PM2.5, persentase hari tinggi, dan indeks gabungan tertinggi. Hasil kedua indikator utama juga dibandingkan untuk mengetahui apakah keduanya menunjukkan stasiun yang sama.

```python
highest_mean_station = station_stats["mean"].idxmax()
highest_high_day_station = station_stats["pct_hari_tinggi"].idxmax()
highest_combined_station = station_stats["indeks_gabungan"].idxmax()

average_combined_index = station_stats["indeks_gabungan"].mean()

print(f"Stasiun dengan rata-rata PM2.5 tertinggi: {highest_mean_station}")
print(f"Stasiun dengan persentase hari tinggi terbesar: {highest_high_day_station}")
print(f"Stasiun dengan indeks gabungan tertinggi: {highest_combined_station}")

print(
    "Kedua indikator menunjukkan stasiun yang sama."
    if highest_mean_station == highest_high_day_station
    else "Kedua indikator menunjukkan stasiun yang berbeda."
)
```

#### Preparing PM10 and Meteorological Variables

Memilih PM10 sebagai variabel target dan WSPM, TEMP, PRES, DEWP, serta RAIN sebagai variabel meteorologis pembanding. Nilai missing dipertahankan dan ditangani secara pairwise deletion saat perhitungan korelasi.

#### Calculating Pearson and Spearman Correlations with PM10

Menghitung koefisien korelasi Pearson dan Spearman antara PM10 dan setiap variabel meteorologis dengan menyertakan jumlah pasangan observasi valid untuk menunjukkan dasar perhitungan setiap koefisien.

```python
pearson_corr = (
    weather.corr(method="pearson")["PM10"]
    .drop("PM10")
)
spearman_corr = (
    weather.corr(method="spearman")["PM10"]
    .drop("PM10")
)

paired_valid_count = (
    weather.notna().astype(int).T
    .dot(weather.notna().astype(int))
    .loc[WEATHER_COLS, "PM10"]
)

correlation_result = pd.DataFrame({
    "Pearson": pearson_corr,
    "Spearman": spearman_corr,
    "Jumlah Pasangan Valid": paired_valid_count
})

correlation_result = correlation_result.reindex(
    correlation_result["Pearson"].abs().sort_values(ascending=False).index
)

display(correlation_result.round(3))
```

Hasil korelasi menunjukkan bahwa WSPM memiliki hubungan negatif paling kuat dengan PM10, baik berdasarkan Pearson (-0,178) maupun Spearman (-0,250), yang mengindikasikan kecenderungan konsentrasi PM10 menurun ketika kecepatan angin meningkat. DEWP menunjukkan hubungan positif, dengan korelasi Spearman (0,149) lebih kuat dibandingkan Pearson (0,068). Sementara itu, TEMP, RAIN, dan PRES memiliki hubungan negatif yang relatif lemah dengan PM10. Secara keseluruhan, seluruh variabel meteorologis menunjukkan korelasi linear maupun monotonik yang lemah terhadap PM10. Analisis menggunakan 419.139 pasangan observasi valid untuk setiap variabel.

#### Identifying Positive, Negative, and Weakest Relationships

Untuk masing-masing metode korelasi, identifikasi hubungan paling positif, paling negatif, dan paling lemah berdasarkan nilai absolut koefisien terkecil. Hal ini melengkapi identifikasi hubungan terkuat berdasarkan nilai absolut terbesar.

```python
for method_name, corr_values in [
    ("Pearson", pearson_corr),
    ("Spearman", spearman_corr)
]:
    strongest_positive = corr_values.idxmax()
    strongest_negative = corr_values.idxmin()
    weakest = corr_values.abs().idxmin()
    strongest_absolute = corr_values.abs().idxmax()

    print(f"\n{method_name}")
    print(
        f"Hubungan positif terbesar: {strongest_positive} "
        f"({corr_values[strongest_positive]:.3f})"
    )
    print(
        f"Hubungan negatif terbesar: {strongest_negative} "
        f"({corr_values[strongest_negative]:.3f})"
    )
    print(
        f"Hubungan terlemah: {weakest} "
        f"({corr_values[weakest]:.3f})"
    )
    print(
        f"Hubungan terkuat secara absolut: {strongest_absolute} "
        f"({corr_values[strongest_absolute]:.3f})"
    )
```

Berdasarkan perbandingan kedua metode, WSPM menunjukkan hubungan negatif paling kuat dengan PM10, baik menggunakan Pearson (-0,178) maupun Spearman (-0,250). Sebaliknya, DEWP memiliki hubungan positif terbesar pada kedua metode, masing-masing 0,068 dan 0,149. Variabel dengan hubungan terlemah berbeda, yaitu PRES pada Pearson (-0,017) dan TEMP pada Spearman (-0,019). Secara keseluruhan, kedua metode menghasilkan pola hubungan yang relatif konsisten, meskipun kekuatan korelasi Spearman lebih besar pada beberapa variabel. Hal ini mengindikasikan bahwa hubungan monotonik dapat lebih terlihat dibandingkan hubungan linear, tetapi seluruh korelasi yang diperoleh tetap tergolong lemah.

#### Comparing Pearson and Spearman Correlation Matrices

Menampilkan heatmap matriks korelasi Pearson dan Spearman untuk seluruh variabel yang dianalisis (PM10, WSPM, TEMP, PRES, DEWP, dan RAIN). Visualisasi digunakan untuk melihat pola hubungan antarvariabel secara keseluruhan, sedangkan tabel korelasi sebelumnya menjadi acuan utama dalam menginterpretasikan hubungan masing-masing variabel dengan PM10.

```python
pearson_matrix = weather.corr(method="pearson")
spearman_matrix = weather.corr(method="spearman")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for ax, matrix, title in zip(
    axes,
    [pearson_matrix, spearman_matrix],
    ["Pearson", "Spearman"]
):
    sns.heatmap(
        matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        linewidths=0.5,
        annot_kws={"size": 11},
        cbar_kws={"shrink": 0.85},
        ax=ax
    )

    ax.set_title(f"Korelasi {title}", fontsize=14)
    ax.tick_params(axis="both", labelsize=11)
    ax.tick_params(axis="x", rotation=45)
    ax.tick_params(axis="y", rotation=0)

fig.suptitle(
    "Perbandingan Korelasi Variabel Meteorologis dengan PM10",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

Berdasarkan heatmap, korelasi Pearson dan Spearman menunjukkan bahwa hubungan antara variabel meteorologis dengan PM10 cenderung lemah. Variabel WSPM memiliki korelasi negatif lemah terhadap PM10, yaitu -0,18 pada Pearson dan -0,25 pada Spearman, yang menunjukkan kecenderungan bahwa peningkatan kecepatan angin berhubungan dengan penurunan konsentrasi PM10. Sementara itu, DEWP menunjukkan korelasi positif, yaitu 0,07 pada Pearson dan 0,15 pada Spearman, sedangkan TEMP, PRES, dan RAIN memiliki korelasi yang sangat lemah dengan PM10. Di sisi lain, terlihat hubungan yang lebih kuat antarvariabel meteorologis, seperti korelasi positif antara TEMP dan DEWP (0,82) serta korelasi negatif antara TEMP dan PRES (-0,81). Secara keseluruhan, hasil ini menunjukkan bahwa kecepatan angin memiliki hubungan linear maupun monotonik yang relatif lebih terlihat dengan PM10 dibandingkan variabel meteorologis lainnya, meskipun kekuatan korelasinya masih tergolong lemah.

Gunakan diagram batang horizontal berpasangan (*grouped horizontal bar chart*) untuk membandingkan nilai dan arah koefisien Pearson dan Spearman antara PM10 dan setiap variabel meteorologis (WSPM, TEMP, PRES, DEWP, dan RAIN). Garis vertikal pada nol menjadi acuan untuk membedakan korelasi positif di sisi kanan dan korelasi negatif di sisi kiri.

```python
fig, ax = plt.subplots(figsize=(11, 6))

correlations = {
    "Pearson": pearson_corr,
    "Spearman": spearman_corr
}

colors = {
    "Pearson": "steelblue",
    "Spearman": "orange"
}

variables_corr = WEATHER_COLS
y = np.arange(len(variables_corr))
bar_height = 0.35

for method_name, corr_values in correlations.items():
    values = corr_values.reindex(variables_corr).values
    offset = -bar_height / 2 if method_name == "Pearson" else bar_height / 2

    bars = ax.barh(
        y + offset,
        values,
        height=bar_height,
        color=colors[method_name],
        label=method_name
    )

    for bar, value in zip(bars, values):
        ax.annotate(
            f"{value:.2f}",
            (
                value,
                bar.get_y() + bar.get_height() / 2
            ),
            xytext=(5 if value >= 0 else -5, 0),
            textcoords="offset points",
            ha="left" if value >= 0 else "right",
            va="center",
            fontsize=11
        )

ax.axvline(0, color="black", linewidth=0.8)

ax.set_yticks(y)
ax.set_yticklabels(variables_corr)
ax.set_xlim(-1.05, 1.05)

ax.set_title(
    "Perbandingan Korelasi Pearson dan Spearman terhadap PM10",
    fontsize=14
)
ax.set_xlabel("Koefisien Korelasi", fontsize=12)
ax.set_ylabel("Variabel Meteorologis", fontsize=12)

ax.tick_params(axis="both", labelsize=12)
ax.legend(title="Metode", fontsize=10, title_fontsize=11)

ax.grid(axis="x", alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

Berdasarkan grafik perbandingan, variabel WSPM (kecepatan angin) menunjukkan korelasi negatif paling kuat terhadap PM10, yaitu -0,18 pada Pearson dan -0,25 pada Spearman. Hal ini mengindikasikan bahwa peningkatan kecepatan angin cenderung berkaitan dengan penurunan konsentrasi PM10. Sementara itu, DEWP memiliki korelasi positif sebesar 0,07 pada Pearson dan 0,15 pada Spearman, yang menunjukkan kecenderungan hubungan positif meskipun sangat lemah. Variabel TEMP, PRES, dan RAIN menunjukkan korelasi yang sangat lemah terhadap PM10, dengan nilai koefisien mendekati nol. #### Visualizing the Relationship Between Meteorological Variables and PM10

Gunakan scatter plot untuk mengamati hubungan antara masing-masing variabel meteorologis dan konsentrasi PM10. Garis regresi linear (regression line) dihitung menggunakan seluruh pasangan data valid, bukan hanya sampel yang ditampilkan. Untuk menghindari pengaruh visual dari nilai ekstrem pada rentang prediktor, garis regresi ditampilkan pada rentang persentil pertama hingga persentil ke-99. Visualisasi ini digunakan untuk mengamati kecenderungan hubungan linear, sedangkan nilai slope dan R² menjadi acuan dalam menginterpretasikan kekuatan hubungan tersebut.

```python
fig, axes = plt.subplots(2, 3, figsize=(17, 10), sharey=True)
axes = axes.flatten()

for i, column in enumerate(WEATHER_COLS):
    ax = axes[i]
    group = weather[[column, "PM10"]].dropna()

    if len(group) < 2 or group[column].nunique() < 2:
        ax.set_visible(False)
        continue

    fit = scipy_stats.linregress(
        group[column],
        group["PM10"]
    )

    sample = group.sample(
        n=min(30000, len(group)),
        random_state=42
    )

    ax.scatter(
        sample[column],
        sample["PM10"],
        s=8,
        alpha=0.18,
        color="steelblue"
    )

    x_values = np.linspace(
        group[column].quantile(0.01),
        group[column].quantile(0.99),
        100
    )

    y_values = fit.intercept + fit.slope * x_values

    ax.plot(
        x_values,
        y_values,
        color="red",
        linewidth=2
    )

    if column in ["WSPM", "RAIN"]:
        ax.set_xlim(left=0)

    ax.set_ylim(bottom=0)

    ax.set_title(
        f"{column} terhadap PM10\n"
        f"Slope = {fit.slope:.3f} | R² = {fit.rvalue ** 2:.3f}",
        fontsize=12
    )

    ax.set_xlabel(column, fontsize=12)
    ax.set_ylabel("PM10 (µg/m³)", fontsize=12)
    ax.tick_params(axis="both", labelsize=11)
    ax.grid(axis="both", alpha=0.3)
    ax.set_axisbelow(True)

for i in range(len(WEATHER_COLS), len(axes)):
    axes[i].set_visible(False)

fig.suptitle(
    "Hubungan Variabel Meteorologis dengan Konsentrasi PM10",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

#### Classifying Monitoring Stations by Average PM2.5 Concentration

Mengelompokkan stasiun pemantauan berdasarkan nilai rata-rata harian PM2.5 (mean_pm25) menggunakan interval kategori yang ditetapkan dalam kode analisis. Kategori yang digunakan terdiri atas Excellent (0–35 µg/m³), Good (>35–75 µg/m³), Slight pollution (>75–115 µg/m³), Moderate pollution (>115–150 µg/m³), Heavy pollution (>150–250 µg/m³), dan Severe pollution (>250 µg/m³). Fungsi kategorikan_pm25() menerapkan batas interval tersebut pada rata-rata harian tiap stasiun. Dengan demikian, kategori yang dihasilkan menggambarkan klasifikasi nilai rata-rata per stasiun, bukan kategori untuk setiap hari atau setiap observasi per jam.

```python
PM25_BINS = [0, 35, 75, 115, 150, 250, np.inf]
PM25_LABELS = [
    "Excellent",
    "Good",
    "Slight pollution",
    "Moderate pollution",
    "Heavy pollution",
    "Severe pollution"
]
PM25_RANGE_LABELS = [
    "0–35",
    ">35–75",
    ">75–115",
    ">115–150",
    ">150–250",
    ">250"
]

PM25_COLORS = {
    "Excellent": "#2E8B57",
    "Good": "#A8D672",
    "Slight pollution": "#F4D03F",
    "Moderate pollution": "#F39C12",
    "Heavy pollution": "#E74C3C",
    "Severe pollution": "#8E44AD"
}

def kategorikan_pm25(data):
    return pd.cut(
        data,
        bins=PM25_BINS,
        labels=PM25_LABELS,
        include_lowest=True,
        right=True
    )

station_geo["kategori_pm25"] = kategorikan_pm25(station_geo["mean_pm25"])

station_geo["kategori_pm25"] = station_geo["kategori_pm25"].cat.reorder_categories(
    PM25_LABELS,
    ordered=True
)

display(
    station_geo[["station", "mean_pm25", "total_hari", "kategori_pm25"]]
    .sort_values("mean_pm25", ascending=False)
    .round(2)
)
```

Tabel menunjukkan hasil klasifikasi berdasarkan rata-rata harian PM2.5 pada masing-masing dari 12 stasiun pemantauan. Dongsi memiliki rata-rata tertinggi sebesar 86,02 µg/m³, sedangkan Dingling memiliki rata-rata terendah sebesar 66,40 µg/m³. Berdasarkan interval kategori yang ditetapkan dalam kode, sembilan stasiun termasuk kategori Slight pollution dan tiga stasiun—Changping, Huairou, serta Dingling—termasuk kategori Good. Sembilan stasiun tersebut meliputi seluruh stasiun Main Urban dan stasiun Sub Urban Shunyi.

Jumlah hari valid pada setiap stasiun berkisar antara 1.438 hingga 1.461 hari. Perlu diingat bahwa jumlah sembilan dan tiga stasiun tersebut merupakan jumlah stasiun dalam masing-masing kategori berdasarkan nilai rata-rata harian per stasiun, bukan jumlah atau persentase seluruh hari pengamatan.
